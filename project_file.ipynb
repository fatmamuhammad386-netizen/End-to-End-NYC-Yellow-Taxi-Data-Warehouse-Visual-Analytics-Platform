```python
# Error Logging Setup
# Initialize an empty DataFrame to store isolated error records with reason tags

df_errors = pd.DataFrame()
```

```python
# Passenger Count Cleaning (passenger_count)
# Fill missing passenger values with the default (1) and convert to integer

df['passenger_count'] = df['passenger_count'].fillna(1).astype(int)
```

```python
# Isolate invalid passenger counts (< 1 or > 6)
invalid_passengers = df[(df["passenger_count"] < 1) | (df["passenger_count"] > 6)].copy()
invalid_passengers["error_reason"] = "Invalid Passenger Count (<1 or >6)"
df_errors = pd.concat([df_errors, invalid_passengers])
df = df.drop(index=invalid_passengers.index)
```

```python
# Payment Type Cleaning (payment_type)
# Replace implicit null value 0 with 5 (Unknown, per NYC Taxi dataset dictionary)
df['payment_type']= df['payment_type'].replace(0,5)
```

```python
# Isolate voided trips (payment_type == 6)
invalid_payments = df[df["payment_type"] == 6].copy()
invalid_payments["error_reason"]="Voided Trip (payment_type == 6)"
# Append to errors DataFrame and drop from main clean DataFrame
df_errors = pd.concat([df_errors ,invalid_payments ])
df = df.drop(index=invalid_payments.index)
```

```python
# Rate Code Cleaning (RatecodeID)
# Fill missing values with default 1 (Standard Rate) and cast to integer

df['RatecodeID']=df['RatecodeID'].fillna(1).astype(int)
```

```python
df['RatecodeID'].value_counts()
```

```python
# Isolate invalid rate codes (values > 6, such as code 99)
invalid_rate=df[df['RatecodeID']>6].copy()
invalid_rate["error_reason"]="Invalid RatecodeID (> 6)"
# Append to errors DataFrame and drop from main clean DataFrame
df_errors=pd.concat([df_errors,invalid_rate])
df=df.drop(index=invalid_rate.index)
```

```python
# Financial Columns Cleaning (fare_amount & total_amount)
# Isolate records with negative or zero fare/total amounts

invalid_amount = df[(df['total_amount']<=0) | (df['fare_amount']<=0)].copy()
invalid_amount["error_reason"]= "Negative or Zero Fare"
# Append to errors DataFrame and drop from main clean DataFrame
df_errors = pd.concat([df_errors , invalid_amount])
df=df.drop(index=invalid_amount.index)
```

```python
# Datetime & Logical Time Validation
# Isolate records where dropoff timestamp is before or equal to pickup timestamp
invalid_time=df[df['tpep_dropoff_datetime']<=df["tpep_pickup_datetime"]].copy()
invalid_time["error_reason"]= "Dropoff before or equal to Pickup time"
df_errors =pd.concat([df_errors , invalid_time])

df=df.drop(index=invalid_time.index)
```

```python
# Trip Distance Cleaning (trip_distance)
# Isolate trips with distance <= 0
invalid_distance= df[df['trip_distance']<=0].copy()
invalid_distance["error_reason"]="Invalid Distance (<= 0)"
df_errors=pd.concat([df_errors,invalid_distance])
df=df.drop(index=invalid_distance.index)
```

```python
# Isolate out-of-range dates (pickups outside of January 2024)
invalid_dates = df[(df["tpep_pickup_datetime"] < "2024-01-01") | (df["tpep_pickup_datetime"] >= "2024-02-01")].copy()
invalid_dates["error_reason"] = "Out of Range Pickup Date (Not Jan 2024)"
df_errors = pd.concat([df_errors, invalid_dates])
df = df.drop(index=invalid_dates.index)
```

```python
# Remaining Null Values Handling
# Impute missing location IDs with 265 (Unknown Location Code) and cast to integer
df["PULocationID"] = df["PULocationID"].fillna(265).astype(int)
df["DOLocationID"] = df["DOLocationID"].fillna(265).astype(int)
# Impute missing VendorID and store_and_fwd_flag with sensible defaults
df["VendorID"] = df["VendorID"].fillna(1).astype(int)
df["store_and_fwd_flag"] = df["store_and_fwd_flag"].fillna("N")
```

```python
# Additional Negative Values Check (Surcharges, Taxes, Tips, Tolls)
# Identify financial columns that should not contain negative values
financial_cols = [
    "extra", "mta_tax", "tip_amount", "tolls_amount", 
    "improvement_surcharge", "congestion_surcharge", "Airport_fee"
]
# Create condition for any negative values in these financial fields
neg_condition = (df[financial_cols] < 0).any(axis=1)
invalid_negatives = df[neg_condition].copy()
invalid_negatives["error_reason"] = "Negative values found in surcharge/tax/tip/toll fields"

df_errors = pd.concat([df_errors, invalid_negatives])
df = df.drop(index=invalid_negatives.index)
```

```python
# Extreme Outliers Filtering (> $5,000)
# Isolate unrealistically high total or fare amounts
extreme_outliers = df[(df["total_amount"] > 5000) | (df["fare_amount"] > 5000)].copy()
extreme_outliers["error_reason"] = "Extreme Outlier Fare/Total Amount (> $5,000)"

df_errors = pd.concat([df_errors, extreme_outliers])
df = df.drop(index=extreme_outliers.index)
```

```python
#  Trip Duration Calculation & Ultra-Short Trips Handling (< 10 seconds)
# Calculate trip duration in seconds for logical validation
trip_duration_seconds = (df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]).dt.total_seconds()

# Isolate trips that lasted less than 10 seconds (Unrealistic / System Glitch)
short_trips = df[trip_duration_seconds < 10].copy()
short_trips["error_reason"] = "Unrealistic Trip Duration (< 10 seconds)"

df_errors = pd.concat([df_errors, short_trips])
df = df.drop(index=short_trips.index)
```

```python
# 13. Create New Feature Column (trip_duration_min)
# Add a clean, calculated feature for trip duration in minutes for downstream analysis
df["trip_duration_min"] = (
    (df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]).dt.total_seconds() / 60.0
).round(2)
```

```python
df["pickup_day_name"] = df["tpep_pickup_datetime"].dt.day_name()
df["is_weekend"] = df["tpep_pickup_datetime"].dt.dayofweek.isin([5, 6]).astype(int)
duration_hours = df["trip_duration_min"] / 60.0
df["avg_speed_mph"] = (df["trip_distance"] / duration_hours).round(2)
```

```python
# Database Connection Configuration
from sqlalchemy import create_engine
SERVER_NAME = "FATMAH"  
DATABASE_NAME = "NYC_Taxi_DW"
connection_string = (
    f"mssql+pyodbc://@{SERVER_NAME}/{DATABASE_NAME}"
    "?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
)
engine = create_engine(connection_string)
```

```python

vendors = pd.DataFrame(
    [
        {"VendorID": 1, "VendorName": "Creative Mobile Technologies, LLC"},
        {"VendorID": 2, "VendorName": "VeriFone Inc."},
    ]
)

payments = pd.DataFrame(
    [
        {"PaymentTypeID": 1, "PaymentDescription": "Credit card"},
        {"PaymentTypeID": 2, "PaymentDescription": "Cash"},
        {"PaymentTypeID": 3, "PaymentDescription": "No charge"},
        {"PaymentTypeID": 4, "PaymentDescription": "Dispute"},
        {"PaymentTypeID": 5, "PaymentDescription": "Unknown"},
        {"PaymentTypeID": 6, "PaymentDescription": "Voided trip"},
    ]
)

ratecodes = pd.DataFrame(
    [
        {"RatecodeID": 1, "RateDescription": "Standard rate"},
        {"RatecodeID": 2, "RateDescription": "JFK"},
        {"RatecodeID": 3, "RateDescription": "Newark"},
        {"RatecodeID": 4, "RateDescription": "Nassau or Westchester"},
        {"RatecodeID": 5, "RateDescription": "Negotiated flat rate"},
        {"RatecodeID": 6, "RateDescription": "Group ride"},
    ]
)

unique_locs = set(df["PULocationID"].dropna().unique()).union(
    set(df["DOLocationID"].dropna().unique())
)
loc_df = pd.DataFrame({"LocationID": [int(x) for x in unique_locs]})

with engine.begin() as conn:
    vendors.to_sql("Dim_Vendor", con=conn, if_exists="append", index=False)
    payments.to_sql(
        "Dim_PaymentType", con=conn, if_exists="append", index=False
    )
    ratecodes.to_sql("Dim_Ratecode", con=conn, if_exists="append", index=False)
    loc_df.to_sql("Dim_Location", con=conn, if_exists="append", index=False)


date_range = pd.date_range(start="2024-01-01", end="2024-12-31", freq="D")
dim_date = pd.DataFrame({"FullDate": date_range})
dim_date["DateKey"] = dim_date["FullDate"].dt.strftime("%Y%m%d").astype(int)
dim_date["Year"] = dim_date["FullDate"].dt.year
dim_date["Quarter"] = dim_date["FullDate"].dt.quarter
dim_date["QuarterName"] = "Q" + dim_date["Quarter"].astype(str)
dim_date["Month"] = dim_date["FullDate"].dt.month
dim_date["MonthName"] = dim_date["FullDate"].dt.month_name()
dim_date["DayOfMonth"] = dim_date["FullDate"].dt.day
dim_date["DayOfWeek"] = dim_date["FullDate"].dt.dayofweek + 1
dim_date["DayName"] = dim_date["FullDate"].dt.day_name()
dim_date["IsWeekend"] = dim_date["DayOfWeek"].isin([6, 7]).astype(int)
dim_date["IsHoliday"] = 0

time_range = pd.date_range(
    start="00:00:00", end="23:59:59", freq="s"
).time
dim_time = pd.DataFrame({"TimeValue": time_range})
dim_time["TimeKey"] = [
    int(t.strftime("%H%M%S")) for t in dim_time["TimeValue"]
]
dim_time["Hour"] = [t.hour for t in dim_time["TimeValue"]]
dim_time["Minute"] = [t.minute for t in dim_time["TimeValue"]]
dim_time["AmPm"] = [t.strftime("%p") for t in dim_time["TimeValue"]]

hours = dim_time["Hour"]
conditions = [
    (hours >= 5) & (hours < 12),
    (hours >= 12) & (hours < 17),
    (hours >= 17) & (hours < 22),
]
choices = ["Morning", "Afternoon", "Evening"]
dim_time["TimeBucket"] = np.select(conditions, choices, default="Night")

with engine.begin() as conn:
    dim_date.to_sql("Dim_Date", con=conn, if_exists="append", index=False)
    dim_time.to_sql("Dim_Time", con=conn, if_exists="append", index=False)

```

```python
vendor_map = pd.read_sql(
    "SELECT VendorKey, VendorID FROM Dim_Vendor", con=engine
).set_index("VendorID")["VendorKey"]
rate_map = pd.read_sql(
    "SELECT RatecodeKey, RatecodeID FROM Dim_Ratecode", con=engine
).set_index("RatecodeID")["RatecodeKey"]
pay_map = pd.read_sql(
    "SELECT PaymentTypeKey, PaymentTypeID FROM Dim_PaymentType", con=engine
).set_index("PaymentTypeID")["PaymentTypeKey"]
loc_map = pd.read_sql(
    "SELECT LocationKey, LocationID FROM Dim_Location", con=engine
).set_index("LocationID")["LocationKey"]

df["VendorKey"] = df["VendorID"].map(vendor_map)
df["RatecodeKey"] = df["RatecodeID"].map(rate_map)
df["PaymentTypeKey"] = df["payment_type"].map(pay_map)
df["PULocationKey"] = df["PULocationID"].map(loc_map)
df["DOLocationKey"] = df["DOLocationID"].map(loc_map)

df["PickupDateKey"] = (
    df["tpep_pickup_datetime"].dt.strftime("%Y%m%d").astype(int)
)
df["DropoffDateKey"] = (
    df["tpep_dropoff_datetime"].dt.strftime("%Y%m%d").astype(int)
)
df["PickupTimeKey"] = (
    df["tpep_pickup_datetime"].dt.strftime("%H%M%S").astype(int)
)
df["DropoffTimeKey"] = (
    df["tpep_dropoff_datetime"].dt.strftime("%H%M%S").astype(int)
)

df.rename(
    columns={
        "store_and_fwd_flag": "StoreAndFwdFlag",
        "passenger_count": "PassengerCount",
        "trip_distance": "TripDistance",
        "fare_amount": "FareAmount",
        "extra": "Extra",
        "mta_tax": "MtaTax",
        "tip_amount": "TipAmount",
        "tolls_amount": "TollsAmount",
        "improvement_surcharge": "ImprovementSurcharge",
        "congestion_surcharge": "CongestionSurcharge",
        "Airport_fee": "AirportFee",
        "total_amount": "TotalAmount",
        "trip_duration_min": "TripDurationMin",
        "avg_speed_mph": "AvgSpeedMph",
    },
    inplace=True,
)

fact_columns = [
    "VendorKey",
    "RatecodeKey",
    "PaymentTypeKey",
    "PULocationKey",
    "DOLocationKey",
    "PickupDateKey",
    "DropoffDateKey",
    "PickupTimeKey",
    "DropoffTimeKey",
    "StoreAndFwdFlag",
    "PassengerCount",
    "TripDistance",
    "FareAmount",
    "Extra",
    "MtaTax",
    "TipAmount",
    "TollsAmount",
    "ImprovementSurcharge",
    "CongestionSurcharge",
    "AirportFee",
    "TotalAmount",
    "TripDurationMin",
    "AvgSpeedMph",
]

drop_cols = [col for col in df.columns if col not in fact_columns]
df.drop(columns=drop_cols, inplace=True)

df.to_sql(
    name="Fact_TaxiTrips",
    con=engine,
    if_exists="append",
    index=False,
    chunksize=50000,
)
```

```python
# Ingest Error Logs into Audit Table
# Append isolated invalid records to Audit_TaxiErrors table
engine.dispose()
with engine.begin() as conn:
    df_errors.to_sql(
        name="Audit_TaxiErrors",
        con=conn,
        if_exists="append",
        index=False,
        chunksize=50000,
    )
```
